## 1. Introduction

Deep Learning is a subset of Machine Learning that uses artificial
neural networks with multiple layers to learn patterns from data.

In this project, PyTorch was used to understand the implementation
and training of deep learning models.

The MNIST handwritten digit dataset was used for image classification.

## 2. PyTorch

PyTorch is an open-source deep learning framework used for building,
training and evaluating neural network models.

Important features used in this project include:

- Tensors
- Neural network modules
- Loss functions
- Optimizers
- Automatic differentiation
- DataLoaders

## 3. Convolutional Neural Network

A Convolutional Neural Network (CNN) is a deep learning architecture
commonly used for image-related tasks.

The CNN developed in this project contains:

- Convolutional layers
- ReLU activation
- Max pooling
- Fully connected layers
- Output layer

The CNN was trained to classify handwritten digits from the MNIST dataset.

## 4. Dataset

The MNIST dataset contains handwritten digits from 0 to 9.

Dataset details:

- Training images: 60,000
- Testing images: 10,000
- Image size: 28 × 28 pixels
- Image type: Grayscale
- Number of classes: 10

## 5. Model Training

The CNN model was trained using the following configuration:

- Loss Function: Cross Entropy Loss
- Optimizer: Adam
- Learning Rate: 0.001
- Batch Size: 64
- Number of Epochs: 5

During training, the loss was monitored to observe how the model
learned from the training dataset.

## 6. Hyperparameter Experimentation

Different hyperparameter configurations were tested to study their
effect on CNN performance.

The parameters experimented with were:

- Learning Rate
- Batch Size
- Number of Epochs

Multiple models were trained using different configurations and
their test accuracy was recorded.

```python
import os
import pandas as pd

# Find the CSV file automatically
csv_path = None

for root, folders, files in os.walk(".."):
    if "hyperparameter_results.csv" in files:
        csv_path = os.path.join(root, "hyperparameter_results.csv")
        break

print("CSV file found at:", csv_path)

# Load the results
results_df = pd.read_csv(csv_path)

results_df
```

```python
best_experiment = results_df.loc[
    results_df["Test Accuracy"].idxmax()
]

print("Best experimental configuration:")
print(best_experiment)
```

```python
best_experiment = results_df.loc[
    results_df["Test Accuracy"].idxmax()
]

print("Best experimental configuration:")
print(best_experiment)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(
    results_df["Experiment"].astype(str),
    results_df["Test Accuracy"]
)

plt.xlabel("Experiment")
plt.ylabel("Test Accuracy (%)")
plt.title("Hyperparameter Experiment Results")

plt.show()
```
