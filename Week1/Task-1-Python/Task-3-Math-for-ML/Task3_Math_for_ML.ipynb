```python
# Task 3 - Math for Machine Learning

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
# Linear Algebra - Vectors

vector_a = np.array([2, 4, 6])
vector_b = np.array([1, 3, 5])

print("Vector A:", vector_a)
print("Vector B:", vector_b)

# Vector addition
addition = vector_a + vector_b
print("Vector Addition:", addition)

# Vector subtraction
subtraction = vector_a - vector_b
print("Vector Subtraction:", subtraction)

# Dot product
dot_product = np.dot(vector_a, vector_b)
print("Dot Product:", dot_product)
```

```python
# Linear Algebra - Matrices

matrix_a = np.array([
    [1, 2],
    [3, 4]
])

matrix_b = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(matrix_a)

print("\nMatrix B:")
print(matrix_b)

# Matrix addition
matrix_addition = matrix_a + matrix_b
print("\nMatrix Addition:")
print(matrix_addition)

# Matrix multiplication
matrix_multiplication = np.dot(matrix_a, matrix_b)
print("\nMatrix Multiplication:")
print(matrix_multiplication)

# Transpose
matrix_transpose = matrix_a.T
print("\nTranspose of Matrix A:")
print(matrix_transpose)
```

```python
# Statistics

data = np.array([10, 20, 30, 40, 50])

mean_value = np.mean(data)
median_value = np.median(data)
variance_value = np.var(data)
standard_deviation = np.std(data)

print("Data:", data)
print("Mean:", mean_value)
print("Median:", median_value)
print("Variance:", variance_value)
print("Standard Deviation:", standard_deviation)
```

```python
# Correlation

data = pd.DataFrame({
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8],
    "Marks": [45, 50, 58, 65, 72, 80, 88]
})

correlation = data["Study_Hours"].corr(data["Marks"])

print("Correlation between Study Hours and Marks:", correlation)
```

```python
# Scatter Plot - Study Hours vs Marks

plt.figure(figsize=(8, 5))

plt.scatter(data["Study_Hours"], data["Marks"])

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.tight_layout()
plt.show()
```

```python
# Probability

total_outcomes = 6
favorable_outcomes = 1

probability = favorable_outcomes / total_outcomes

print("Total outcomes:", total_outcomes)
print("Favorable outcomes:", favorable_outcomes)
print("Probability of getting a specific number:", probability)
```

```python
# Expected Value

outcomes = np.array([10, 20, 30])
probabilities = np.array([0.2, 0.5, 0.3])

expected_value = np.sum(outcomes * probabilities)

print("Expected Value:", expected_value)
```

```python
# Calculus - Derivative

def function(x):
    return x ** 2

def derivative(x):
    return 2 * x

x = 5

print("Function value at x =", x, ":", function(x))
print("Derivative at x =", x, ":", derivative(x))
```

```python
# Gradient Descent

x = 10
learning_rate = 0.1

for i in range(10):
    gradient = 2 * x
    x = x - learning_rate * gradient

print("Final value of x:", x)
```

```python
# Gradient Descent Visualization

x = 10
learning_rate = 0.1

x_values = []
y_values = []

for i in range(10):
    x_values.append(x)
    y_values.append(x ** 2)

    gradient = 2 * x
    x = x - learning_rate * gradient

plt.figure(figsize=(8, 5))

plt.plot(x_values, y_values, marker="o")

plt.title("Gradient Descent Visualization")
plt.xlabel("X")
plt.ylabel("f(x) = x²")

plt.tight_layout()
plt.show()
```
